```python
import seaborn as sns 
import pandas as pd
titanic = sns.load_dataset("titanic")

titanic.shape

titanic.dtypes

titanic.head().T

titanic.nunique()

titanic.isna().sum()

pd.crosstab(titanic["class"], titanic["embark_town"])
```

 3 constats sur la répartition des passagers

1. Southampton est le port d'embarquement le plus fréquent** pour les trois classes, avec notamment 353 passagers de troisième classe.
2. La troisième classe est la plus représentée**, avec 491 passagers au total dans le jeu de données.
3. Les passagers de première classe sont particulièrement nombreux à Cherbourg** : 85 passagers, contre 66 en troisième classe et 17 en deuxième classe.
